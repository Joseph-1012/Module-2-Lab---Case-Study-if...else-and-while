```python
#Joseph Seiwert
#File name: GPA checker
#This app is designed to check whether any student's GPA succeeds a certain 
#value to determine if they have made the Dean's list or Honor roll

while True :
    studentName = input("Input student name (or ZZZ to quit): ")
    
    if studentName == "ZZZ" :
        break
        
    studentGPA = float(input("Input student GPA: "))
    
    if studentGPA >= 3.5 :
        print("Student",studentName,"has made the Dean's list.")
        
    elif studentGPA >= 3.25 :
        print("Student",studentName,"has made the Honor Roll.")
        
print("Program terminated")
```
